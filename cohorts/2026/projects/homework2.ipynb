```python
import re
def classify_company(name: str) -> str:
    if not isinstance(name, str):
        return "Other"
    if re.search(r"\bTechnologies\b", name):
        return "Technologies"
    elif re.search(r"\b(Acquisition Corp|Acquisition Corporation|Corp)\b", name):
        return "Acquisition Corp"
    elif re.search(r"\b(Inc|Incorporated)\b", name):
        return "Inc."
    elif re.search(r"\bGroup\b", name):
        return "Group"
    elif re.search(r"\b(Ltd|Limited)\b", name):
        return "Limited"
    elif re.search(r"\b(Holdings|Holding)\b", name):
        return "Holdings"
    else:
        return "Other"

# Apply to your DataFrame
withdrawed_ipo["Company Type"] = withdrawed_ipo["Company"].apply(classify_company)
withdrawed_ipo.head()
```

```python
def parse_price(val):
    if pd.isna(val) or str(val).strip() in ["-", ""]:
        return np.nan
    clean_val = re.sub(r"[^\d.]", "", str(val))
    try:
        return float(clean_val)
    except ValueError:
        return np.nan
withdrawed_ipo["Price_Low"] = withdrawed_ipo["Price_Low"].apply(parse_price)
withdrawed_ipo["Price_High"] = withdrawed_ipo["Price_High"].apply(parse_price)
withdrawed_ipo["Avg_price"] = withdrawed_ipo[["Price_Low", "Price_High"]].mean(axis=1)
withdrawed_ipo.head()
```

```python
withdrawed_ipo["Est_$_Vol_(millions)"] = withdrawed_ipo["Est_$_Vol_(millions)"].apply(parse_price)
withdrawed_ipo.head()
```

```python
# shares_offered = withdrawed_ipo["Shares_(millions)"] * withdrawed_ipo["Est_$_Vol_(millions)"]
withdrawed_ipo["Shares_offered_value"] = withdrawed_ipo["Shares_(millions)"] * withdrawed_ipo["Avg_price"]
withdrawed_ipo["Shares_offered_value"] = withdrawed_ipo["Shares_offered_value"].fillna(withdrawed_ipo["Est_$_Vol_(millions)"])
withdrawed_ipo.head()
```

```python
url = "https://www.iposcoop.com/2025-pricings/"
tables = pd.read_html(url)
ipos_df = tables[0]
ipos_df['Offer Date'] = pd.to_datetime(ipos_df['Offer Date'], errors='coerce')

filtered_ipos = ipos_df[(ipos_df['Offer Date'] < '2025-09-01') & (ipos_df['Return'] != '0.00%')]
filtered_ipos
```

```python
import yfinance as yf
tickers = filtered_ipos['Symbol'].dropna().astype(str).str.strip().tolist()

# 3. Data Download via yfinance
data = yf.download(tickers, start="2025-01-01", end="2026-09-12", group_by='ticker')
data
```

```python
processed_dfs = []
for ticker in tickers:
    try:
        ticker_data = data[ticker] if len(tickers) > 1 else data
        if not ticker_data['Close'].dropna().empty:
            df = ticker_data[['Close']].copy().dropna()
            df = df.reset_index()  # Ensures 'Date' becomes an explicit column
            df['Symbol'] = ticker

            # Feature Engineering: Historical Growth & Sharpe
            df['growth_252d'] = df['Close'] / df['Close'].shift(252)
            df['volatility'] = df['Close'].rolling(30).std() * np.sqrt(252)
            df['Sharpe'] = (df['growth_252d'] - 0.05) / df['volatility']

            processed_dfs.append(df)
    except Exception:
        continue
stocks_df = pd.concat(processed_dfs,ignore_index=True)
stocks_df['Date'] = pd.to_datetime(stocks_df['Date'])
sept_11_df = stocks_df[stocks_df['Date'] == '2026-09-11']
print(f"Median Sharpe on Sep 11, 2026: {sept_11_df['Sharpe'].median():.4f}")
```

```python
stocks_df = stocks_df.sort_values(['Symbol', 'Date']).reset_index(drop=True)

for m in range(1, 13):
    days = m * 21
    # Forward shift grouped by Symbol
    stocks_df[f'future_growth_{m}_m'] = (
        stocks_df.groupby('Symbol')['Close'].shift(-days) / stocks_df['Close']
    )



```

```python
entry_dates = stocks_df.groupby('Symbol')['Date'].min().reset_index()
entry_dates.rename(columns={'Date': 'min_date'}, inplace=True)


ipo_entry_df = pd.merge(
    stocks_df,
    entry_dates,
    left_on=['Symbol', 'Date'],
    right_on=['Symbol', 'min_date'],
    how='inner'
)

growth_cols = [f'future_growth_{m}_m' for m in range(1, 13)]
summary_stats = ipo_entry_df[growth_cols].describe().T

summary_stats.index = [f'{m} Month(s)' for m in range(1, 13)]
print("\nPost-IPO Future Growth Statistics from Min Date")
print(summary_stats[['count', 'mean', '50%', 'std', 'min', 'max']])
peak_month = summary_stats['50%'].idxmax()
peak_median_value = summary_stats['50%'].max()
print(f"\nHighest Median Growth Month: {peak_month} (Median: {peak_median_value:.4f})")
```

```python
tickers_set['is_oversold'] = tickers_set['rsi'] < 30
tickers_set.head()
```

```python
start_date = '2000-01-01'
end_date = '2025-06-01'

filtered_trades = tickers_set[
    (tickers_set['Date'] >= start_date) &
    (tickers_set['Date'] <= end_date) &
    (tickers_set['is_oversold']) &
    (tickers_set['growth_future_30d'].notna())
].copy()
filtered_trades
```

```python
INVESTMENT_PER_TRADE = 1000

filtered_trades['trade_net_income'] = INVESTMENT_PER_TRADE * (filtered_trades['growth_future_30d'] - 1.0)

net_income = filtered_trades['trade_net_income'].sum()

total_trades = len(filtered_trades)
winning_trades = (filtered_trades['trade_net_income'] > 0).sum()
win_rate = (winning_trades / total_trades) * 100 if total_trades > 0 else 0.0

print(f" Strategy Execution Summary ({start_date} to {end_date}) ")
print(f"Total Oversold Signals Executed: {total_trades:,}")
print(f"Winning Trades (> $0 profit): {winning_trades:,} ({win_rate:.2f}%)")
print(f"Average Profit per Trade: ${filtered_trades['trade_net_income'].mean():.2f}")
print(f"Net Cumulative Income: ${net_income:,.2f}")
```
